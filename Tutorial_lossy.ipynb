***Lossy simulation***
===========

Short story long, in simple terms, qubits go missing all the time. When that happens, those whom they are maximally entangled with and left behind will act in a **mix**ture of **states**!

Why does that happen?

Sometimes the route is too long, some times it is too rough, and sometimes both!


In other terms, attenuation exists! Check [Beer-Lambert law](https://en.wikipedia.org/wiki/Beer%E2%80%93Lambert_law) for further information about attenuation and transmittence.

Now let's talk simulation! In this simulator, you can decide how much of your qubits should go bye-bye for a link that connects two components. Let's make a setup first!

![The setup](./Source.png "The setup")

```python
from Scheduler import*
from Measurer import*
from Source import*
from Auto_connector import*
from Classic_data_reader import*
from Detector_module import*
import numpy as np
```

```python
source = Source(1, 1, 2)
measurer_1 = Measuring_Device(1, 0)
measurer_2 = Measuring_Device(2, 0)
detector_module = Detector_module(1)
reader = Classic_data_reader(1)
```

So now we are going to connect these components. The connection function accepts a parameter for delay, probability of loss, and probability of depolarisation for the link object it automatically defines.

```python

connect(detector_module.output_a_1, measurer_1.ch_in_q)
connect(detector_module.output_a_2, measurer_2.ch_in_q)
connect(measurer_1.ch_out_result, reader.input_1)
connect(measurer_2.ch_out_result, reader.input_2)
connect(source.out_ch_1, detector_module.input_a_1, 1, .5)
connect(source.out_ch_2, detector_module.input_a_2, 1, .5)
scheduler.run(10000)

print(f"Total number of qubits on side 1 and 2: {detector_module.Total_qubits_1},{detector_module.Total_qubits_2}")
print(f"The number of Ebits arriving: {len(detector_module.list_a_1)}, {len(detector_module.list_a_2)} \nThe number single qubits missing on: {detector_module.single_qubits_1 + detector_module.single_qubits_2}")
print(f"Number of qubits arrived in different states: 00:{reader.count_00}, 01:{reader.count_01}, 10:{reader.count_10}, 11:{reader.count_11}")


    
```

**Results**

In this tutorial, we see a simple lossy setup. The goal is to learn how to use the loss and see its effects. As the qubits have a 50% chance of arrival, you can see that around 25% of bell states arrive. Around 50% of qubits arrived missing their entangled qubit.
